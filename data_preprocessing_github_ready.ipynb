# Machine Learning Data Preprocessing Using Python

**Sample dataset:** Student Performance

This notebook implements the preprocessing workflow documented in the project report: data loading, inspection, duplicate removal, missing-value handling, feature selection, categorical encoding, Min-Max normalization, and basic exploratory data analysis (EDA).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load the dataset

The raw student-performance dataset is stored as a CSV file and loaded into a Pandas DataFrame.


```python
df = pd.read_csv("student_performance.csv")
df.head()
```

## 2. Initial data inspection

Inspect the dataset size, data types, missing values, and descriptive statistics before transformation.


```python
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDescriptive statistics:")
print(df.describe(include="all"))
```

## 3. Remove duplicate records

The report identifies a repeated record for `Student_ID` 111. Duplicate rows are removed so repeated observations do not receive excessive importance.


```python
print("Duplicate rows before removal:", df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicate rows after removal:", df.duplicated().sum())
print("Shape after duplicate removal:", df.shape)
```

## 4. Handle missing values

Numerical missing values are filled with their median. The missing categorical value in `Gender` is filled with the mode, following the methodology in the report.


```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].median())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

print(df.isnull().sum())
```

## 5. Feature selection

`Student_ID` is an identifier rather than a meaningful predictive feature, so it is excluded. `City` is also excluded from the modelling-ready example, keeping the requested categorical encoding focused on `Gender` and `Internet_Access`.


```python
selected_features = [
    "Age", "Gender", "Study_Hours",
    "Attendance", "Internet_Access", "Final_Score"
]
model_data = df[selected_features].copy()
model_data.head()
```

## 6. Encode categorical variables

One-hot encoding converts `Gender` and `Internet_Access` into numerical binary columns.


```python
encoded = pd.get_dummies(
    model_data,
    columns=["Gender", "Internet_Access"],
    drop_first=True,
    dtype=int
)
encoded.head()
```

## 7. Normalize numerical features

Min-Max normalization scales `Age`, `Study_Hours`, and `Attendance` to the 0–1 range.

Formula:

`x' = (x - min(x)) / (max(x) - min(x))`


```python
scale_cols = ["Age", "Study_Hours", "Attendance"]

for col in scale_cols:
    min_v = encoded[col].min()
    max_v = encoded[col].max()
    encoded[col] = (encoded[col] - min_v) / (max_v - min_v)

encoded.head()
```

## 8. Exploratory Data Analysis (EDA)

Pandas is used for descriptive statistics and correlation analysis. NumPy is used for numerical computing as part of the preprocessing workflow.


```python
numeric_cols = ["Age", "Study_Hours", "Attendance", "Final_Score"]
print(df[numeric_cols].describe())

print("\nCorrelation matrix:")
print(df[["Study_Hours", "Attendance", "Final_Score"]].corr())
```

```python
df[numeric_cols].hist(figsize=(10, 8))
plt.tight_layout()
plt.show()
```

### EDA interpretation

In this small sample, `Study_Hours` and `Attendance` show positive relationships with `Final_Score`. The report notes that these observations should not be treated as definitive because the dataset is small.


## 9. Save the cleaned dataset

The final modelling-ready dataset is saved as a CSV file.


```python
cleaned_csv = "cleaned_student_performance.csv"
encoded.to_csv(cleaned_csv, index=False)

print("Saved:", cleaned_csv)
print("\nFinal missing-value check:")
print(encoded.isnull().sum())
```

## 10. Preprocessing summary

| Step | Action |
|---|---|
| Data loading | Loaded CSV data into a Pandas DataFrame |
| Initial inspection | Checked shape, types, null values and statistics |
| Duplicate removal | Removed the repeated record |
| Missing-value handling | Median for numerical columns; mode for Gender |
| Feature selection | Removed identifier and selected relevant modelling features |
| Categorical encoding | Applied one-hot encoding |
| Normalization | Scaled Age, Study_Hours and Attendance to 0–1 |
| EDA | Calculated descriptive statistics and correlations |

The resulting dataset is a clean, numerical, modelling-ready representation of the selected features.
